3. Explain the following types of operators in Python with suitable examples for Arithmetic operators, Comparison operators, Logical operators, Assignment operators, Membership operators, Identity operators. Also explain the difference between implicit and explicit type conversion.

- . Arithmetic Operators- Used to perform standard mathematical calculations.
•	Examples: + (Addition), - (Subtraction), * (Multiplication), / (Division), % (Modulus/Remainder), ** (Exponent/Power), // (Floor Division).
•	Example:
a = 11
b = 3
print(a % b)   # Output: 2 (Remainder of 11 divided by 3)
print(a // b)  # Output: 3 (Rounds down the division result)

-  Comparison (Relational) Operators- Used to compare two values. The result is always a Boolean value (True or False).
•	Examples: == (Equal to), != (Not equal to), >, <, >=, <=.
•	 Example:
marks = 75
print(marks >= 35)  # Output: True
print(marks == 100) # Output: False

-  Logical Operators- Used to combine conditional statements.
•	and: Returns True if both conditions are true.
•	or: Returns True if at least one condition is true.
•	not: Reverses the logical state (turns True to False and vice versa).
•	Example:
has_hall_ticket = True
is_late = False
print(has_hall_ticket and not is_late)  # Output: True (Allowed to enter exam)
-  Assignment Operators- Used to assign values to variables. Shortcut versions combine arithmetic operations with assignment.
•	Examples: =, +=, -=, *=, /=.
•	Example:
score = 10
score += 5   # Shortcut for score = score + 5
print(score) # Output: 15

-  Membership Operators- Used to test if a value or sequence is present inside a collection like a string, list, or tuple.
•	in: Returns True if the value is found.
•	not in: Returns True if the value is not found.
•	Example:
passed_roll_nos = [5, 12, 18, 24]
print(12 in passed_roll_nos)      # Output: True
print(7 in passed_roll_nos)       # Output: False

-  Identity Operators- Used to check if two variables point to the exact same object in the computer memory.
•	is: Returns True if both variables are the same object.
•	is not: Returns True if they are different objects.
•	Example:
x = [1, 2]
y = [1, 2]
z = x
print(x is z)  # Output: True (Points to the same list memory)
print(x is y)  # Output: False (They have the same data, but are separate objects)


Type conversion is changing a variable from one data type to another:
•	Implicit Type Conversion
Done by: Python automatically.
User effort: No manual code or function is required.
Purpose: Prevents data loss during calculations.
Example: Adding 5 (int) and 2.0 (float) automatically results in 7.0 (float).

•	Explicit Type Conversion
Done by: The programmer manually.
User effort: Requires functions like int(), float(), or str().
Purpose: Forces a specific data type to fix errors or run custom logic.
Example: Converting text "20" to a number using int("20") to perform calculations.

         


4. What is a string in Python? Explain string indexing, slicing, concatenation, repetition and formatting with suitable examples                                                  
In Python, a string is a sequence of characters used to represent text. Strings are created by putting characters inside single quotes ('...'), double quotes ("..."), or triple quotes ("""...""").
Strings are immutable, meaning you cannot change individual characters after the string is created.
String Operations
A. Indexing
Indexing is used to get a single character from a string using its position number inside square brackets [].
•	Positive Indexing: Starts at 0 (first character) and moves right.
•	Negative Indexing: Starts at -1 (last character) and moves left.
•	Example:
text = "Python"
print(text[0])   # Output: P
print(text[4])   # Output: o
print(text[-1])  # Output: n (last letter)

B. Slicing
Slicing extracts a part of a string using the syntax string[start:end:step].
•	start: Where to begin (inclusive).
•	end: Where to stop (exclusive).
•	step: Number of characters to skip forward (default is 1).
EXAMPLE:
text = "Programming"
print(text[0:3])   # Output: Pro
print(text[:4])    # Output: Prog (starts from 0)
print(text[7:])    # Output: ming (goes to the end)
print(text[::-1])  # Output: gnimmargorP (reverses string)

C. Concatenation
Concatenation means joining two or more strings together using the + operator.
EXAMPE:
str1 = "Hello"
str2 = "World"
result = str1 + " " + str2
print(result)  # Output: Hello World
D. Repetition
Repetition repeats a string a specific number of times using the * operator.
EXAMple:
text = "Go!"
print(text * 3)  # Output: Go!Go!Go!
E. Formatting
Formatting lets you inject variables directly into a string template.
1.	f-Strings: Put an f before the string and place variables inside {}.
2.	.format(): Use {} as placeholders and pass variables into the function.
python
name = "Sam"
age = 20

# f-string
print(f"{name} is {age} years old.")  # Output: Sam is 20 years old.

# .format()
print("{} is {} years old.".format(name, age))  # Output: Sam is 20 years old.

A list in Python is a built-in data structure used to store an ordered collection of items. Lists are defined by placing comma-separated values inside square brackets . They can hold elements of different data types (integers, strings, floats, or even other lists) and allow duplicate values.

```python
# 8. Create the following list: marks = [72, 85, 60, 91, 78] Write a Python program to: 1. Add 88 to the list. 2. Remove 60. 3. Sort the list. 4. Display the highest mark. 5. Display the lowest mark. 6. Calculate the average mark.

# Create the initial list of marks
marks = [72, 85, 60, 91, 78]
 # 1. Add 88 to the list
marks.append(88)
# 2. Remove 60
marks.remove(60)
 # 3. Sort the list
marks.sort()
# 4. Display the highest mark
highest = max(marks)
# 5. Display the lowest mark
lowest = min(marks)
# 6. Calculate the average mark
average = sum(marks) / len(marks)
 # Display all results clearly
print("Updated and sorted list:", marks)
print("Highest mark:", highest)
print("Lowest mark:", lowest)
print("Average mark:", average)

```

```python
#9: Consider the following string: "Python Java C C++ Python Java Python" Write a Python program to: 1. Convert the string into a list of programming languages. 2. Display the list. 3. Count how many times "Python" appears. 4. Remove duplicate language names. 5. Join the unique language names using " | "
# Create the initial string
text = "Python Java C C++ Python Java Python"
 # 1. Convert the string into a list of programming languages
languages = text.split()
# 2. Display the list
print("2. List of languages:", languages)
# 3. Count how many times "Python" appears
python_count = languages.count("Python")
print("3. 'Python' count:", python_count)
 # 4. Remove duplicate language names
# Converting to a dict keys preserves the original order, unlike set()
unique_languages = list(dict.fromkeys(languages))
print("4. Unique languages:", unique_languages)
 # 5. Join the unique language names using " | "
joined_string = " | ".join(unique_languages)
print("5. Joined string:", joined_string)

```
