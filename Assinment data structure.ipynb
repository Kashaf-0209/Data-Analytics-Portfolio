 Q3. what is a set in Python? Explain why duplicate values are not stored in a set.Describe union, intersection, and difference with suitable examples.


---



--> A set in Python is an unordered collection of unique elements. It automatically removes duplicates and supports powerful mathematical operations such as union, intersection, and difference.

- A set does not store duplicate values because it is designed to model the mathematical concept of a set, where each element must be unique


*suppose two sets A & B*
```
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
```

**1. Union — Combine All Unique Elements**






````python
A= {1, 2, 3, 4}
B= {3, 4, 5, 6}

# Example 1 (Symbol)

print("Union:", A | B)

# Example 2 (Method)

print("Union (method):", A.union(B))
````

2. **Intersection — Common Elements Only**

````python
# Example 1

print("Intersection:", A & B)

# Example 2

print("Intersection (method):", A.intersection(B))

# Common items → `{3, 4}`
````

**3. Difference — Items in A but Not in B**

````python
# Example 1

print("A - B:", A - B)

# Example 2

print("B - A:", B.difference(A))
````

**4. Symmetric Difference — Elements Not Common to Both**

````python
# Example 1

print("Symmetric Difference:", A ^ B)

# Example 2

print("Symmetric Difference (method):", A.symmetric_difference(B))
````

Q4. What are conditional statements in Python? Explain if, if-else, if-elif-else, and nested
if statements with suitable examples

---

--> Conditional statements in Python are control flow structures that allow a program to make decisions and execute different blocks of code based on whether a condition evaluates to `True` or `False`
- Python relies on indentation (typically 4 spaces) rather than curly braces to define the scope and blocks of these statements.

1. The if Statement:
 The if statement is the simplest form of decision-making. It executes a block of code only if the specified condition is True. If the condition is False, the code block is skipped.
 - Syntax:
      - if condition:
      - code to execute if condition is True

```
Example:age = 20
if age >= 18:
    print("You are eligible to vote.")
```


2. The if-else Statement:
 The if-else statement provides an alternate path of execution. If the if condition evaluates to True, the if block runs. If it evaluates to False, the else block runs instead.
  - Syntax:
    - if condition:
    - Code if condition is True
    - else:
    -  Code if condition is False

```
Example:number = -5
if number >= 0:
    print("The number is positive.")
else:
    print("The number is negative.")
```
3. The if-elif-else Statement: When you need to check multiple conditions sequentially, you use the elif (short for else if) ladder. Python evaluates the conditions from top to bottom. As soon as it finds one that is True, it executes that specific block and skips the rest. If none of the conditions are true, the final else block executes.
- Syntax:
  - if condition1:
  - Code if condition1 is True

```
Example:
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```
4. Nested if Statements: A nested if statement occurs when an if, if-else, or if-elif-else structure is placed inside another if or else block. This is used when a secondary decision depends entirely on the outcome of a primary decision.
- Syntax:
- if outer_condition:
    - Executes if outer_condition is True
- if inner_condition:
   - Executes if both outer and inner conditions are True
```
Example: has_ticket = True
age = 15

if has_ticket:
    print("Welcome to the theater!")
    # Nested condition inside the main 'if'
    if age < 12:
        print("Ticket price: $5 (Child Discount)")
    else:
        print("Ticket price: $10 (Standard)")
else:
    print("Please purchase a ticket at the front desk.")
 ```

Question 5: Explain the difference between for and while loops. Also explain the
purpose of break, continue, pass, and the else clause with loops.

---



--> The primary difference between a for loop and a while loop lies in how the number of iterations is determined. A for loop is definite and iterates over a pre-defined sequence, whereas a while loop is indefinite and runs as long as a specified condition remains true.
- A for loop is used when you know in advance how many times you want to run the code. It automatically steps through a set sequence, like a list of items or a specific range of numbers, and stops when it reaches the end of that sequence.
- A while loop is used when you do not know the exact number of repetitions beforehand. It relies on a true/false condition and will keep running continuously as long as that condition remains true. It only stops when the condition changes to false.

`break`: Terminates the loop entirely. Program execution immediately jumps to the first line of code following the loop block. It is commonly used to exit early when a search or validation condition is met.

`Continue`: Skips the rest of the current iteration and forces the loop to jump directly to the next cycle. Unlike break, it does not exit the loop; it just bypasses the remaining code block for that specific pass.

`pass`: Acts as a null operation or syntactic placeholder. Because Python uses indentation rather than brackets to define code blocks, an empty block will cause a syntax error. pass is used when a statement is required by syntax but you want no action to occur (e.g., when scaffolding code layout).

`else clause`: Executes only if the loop finishes normally. For a for loop, this means it completed all iterations of the sequence. For a while loop, this means the condition naturally turned False. Crucially, if the loop is exited prematurely via a break statement, the else block is completely skipped

4. Remove duplicate language names.
5. Join the unique language names using " | ".

````python
#string:
strng= "Python Java C C++ Python Java Python"

# Convert the string into a list of programming languages.
lst= strng.split()

# Display the list.
print(lst)

# Count how many times "Python" appears.
strng.count("Python")

# Remove duplicate language names.
set_dup= set(lst)
print(set_dup)

# Join the unique language names using " | ".
updated_list = list(set_dup)
print("updated_list:\t", updated_list)
updated_strng= " | ".join(updated_list)
print("updated string:\t", updated_strng)
````

Question 10: Write a Python program to print multiplication tables from 1 to 5.    

Use nested loops to display each table from 1 × 1 up to 5 × 10.

Also modify the program so that:

- The number 5 is skipped using continue.
- The inner loop stops when the multiplier reaches 8 using break

---




````python
# nested loops to display each table from 1 × 1 up to 5 × 10.
for i in range(1,6):
  n=1
  print(f"Table of {i}:")
  while n<11:
    mul=i*n
    print(f"{i} x {n} =",mul)
    n=n+1

print("______________________________________________________________________\n")
print("======================================================================")
print("______________________________________________________________________")


#modified program, so that:
# The number 5 is skipped using continue.
# The inner loop stops when the multiplier reaches 8 using break

for i in range(1,6):
  if i==5:
    continue
  n=1
  print(f"Table of {i}:")
  while n<11:
    mul=i*n
    print(f"{i} x {n} =",mul)
    if n==8:
      break
    n=n+1
````
